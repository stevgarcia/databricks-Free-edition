in this notebook we are going to test the function of foreachbatch taking a table sourcetable and then with a batch function we define two sink destinations using only one streaming process

```python
from pyspark.sql.functions import *
df=spark.readStream.table("workspace.streaming.sourcetable")
```

```python
def myfunction(df,batch_id):
    df=df.groupBy("color").agg(count("*").alias("count"))

    #destination 1

    df.write.format("delta")\
    .mode("append")\
    .option("path","/Volumes/workspace/streaming/stream_volume/foreachsink/destination1/")\
    .save()


    #destination 2

    df.write.format("delta")\
    .mode("append")\
    .option("path","/Volumes/workspace/streaming/stream_volume/foreachsink/destination2/")\
    .save()
```

```python
df.writeStream.foreachBatch(myfunction)\
.outputMode("append")\
.trigger(once=True)\
.option("checkpointLocation","/Volumes/workspace/streaming/stream_volume/foreachsink/checkpoint")\
.start()
```
