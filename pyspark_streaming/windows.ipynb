```python
%sql
CREATE TABLE workspace.streaming.windonwtbl
(
  color STRING,
  event_date TIMESTAMP


)
```

```python
%sql
INSERT INTO workspace.streaming.windonwtbl
VALUES
('red','2025-01-01T11:01:00.000+00:00'),
('green','2025-01-01T11:01:00.000+00:00')
```

```python
df=spark.readStream.table("workspace.streaming.windonwtbl")
```

```python
from pyspark.sql.functions  import *
df=df.groupBy("color",window("event_date","10 minutes")).agg(count(lit(1)).alias("color_count"))
```

```python
df.writeStream.format("delta")\
    .outputMode("complete")\
    .trigger(once=True)\
    .option("path","/Volumes/workspace/streaming/stream_volume/windows/data")\
    .option("checkpointLocation","/Volumes/workspace/streaming/stream_volume/windows/checkpoint")\
    .start()
```
